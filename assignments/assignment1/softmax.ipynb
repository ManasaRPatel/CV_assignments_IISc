**Inline Question 2**

Although gradcheck is reliable softmax loss, it is possible that for SVM loss, once in a while, a dimension in the gradcheck will not match exactly. What could such a discrepancy be caused by? Is it a reason for concern? What is a simple example in one dimension where a svm loss gradient check could fail? How would change the margin affect of the frequency of this happening?

Note that SVM loss for a sample $(x_i, y_i)$ is defined as: $$L_i = \sum_{j\ne y_i}\max(0, s_j - s_{y_i} + \Delta)$$ where $j$ iterates over all classes except the correct class $y_i$ and $s_j$ denotes the classifier score for $j^{th}$ class. $\Delta$ is a scalar margin. For more information, refer to 'Multiclass Support Vector Machine loss' on [this](https://cs231n.github.io/linear-classify/) page.

*Hint: the SVM loss function is not strictly speaking differentiable.*


$\color{blue}{\textit Your Answer:}$
The SVM loss function `max(0, s_j - s_yi + Δ)` is not differentiable at the point where `s_j - s_yi + Δ = 0`. At this point, the gradient is not uniquely defined, and only a subgradient exists.Because of this, the analytical gradient (calculated using one branch of the function) may not match the numerical gradient (calculated using a small interval around the point).This is usually not a problem in practice because such points are very rare. They form a set of measure zero, meaning they are extremely unlikely to occur randomly. This is simply a property of the SVM loss function, which is not smooth everywhere.
For example, consider a simple one-dimensional case with `Δ = 1`, where the other-class score is `w` and the correct-class score is `0`. The loss becomes:
`L = max(0, w - 1)`

At `w = 1`, the loss has a sharp corner (kink), where the gradient is not uniquely defined. Therefore, a gradient check around this point may show a mismatch.
When the margin `Δ` increases, the kink moves to the point `s_j - s_yi = -Δ`. This makes it more likely that some score differences will fall exactly on or very close to a kink during gradient checking. As a result, gradient-check mismatches may occur more frequently with larger margins.

**Inline Question 4** - *True or False*

Suppose the overall training loss is defined as the sum of the per-datapoint loss over all training examples. It is possible to add a new datapoint to a training set that would change the softmax loss, but leave the SVM loss unchanged.

$\color{blue}{\textit Your Answer:}$
True

$\color{blue}{\textit Your Explanation:}$  The Softmax (cross-entropy) loss for a data point is `-log(p_yi)`, which is always greater than zero because the probability of the correct class is less than 1. Therefore, adding a new data point always increases the total Softmax loss.
On the other hand, the SVM (hinge) loss for a new data point can be zero if the correct class is predicted with a margin greater than `Δ` (i.e., `s_yi - s_j > Δ` for all `j ≠ yi`). In this case, adding an easy, correctly classified data point changes the Softmax loss but does not affect the SVM loss.
This is because W is kept fixed, and we are only calculating the total loss again by adding the loss of the new data point.
